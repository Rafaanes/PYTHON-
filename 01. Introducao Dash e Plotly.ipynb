# O que é o Dash?

Dash é um framework (estrutura) em Python que permite criar aplicações web interativas com foco em visualização de dados.

Ele foi desenvolvido pela mesma empresa do Plotly, e é muito usado em ciência de dados, engenharia e negócios para criar dashboards (painéis interativos) que rodam no navegador, sem precisar programar em HTML, CSS ou JavaScript.

# Características principais do Dash:

Interface Web, você cria uma página da web com Python.

Foco em gráficos	ideal para relatórios visuais, dados e análises.

Componentes prontos botões, gráficos, filtros, tabelas, etc. já vêm prontos para usar.

Interatividade os gráficos podem reagir quando você clica, escolhe opções, filtra dados, etc.

🌐 Roda no navegador você roda um servidor local e acessa no navegador (localhost).

-----

# O que é o Plotly?

Plotly é uma biblioteca de gráficos em Python (também disponível para R, JavaScript e outras linguagens). Ela permite criar gráficos bonitos, interativos e profissionais com poucas linhas de código.

# Por que usar Plotly?
Mais interativo que matplotlib ou seaborn.

Você pode passar o mouse sobre os pontos, dar zoom, e clicar nos elementos.

Fácil de integrar com Dash, pois as duas bibliotecas foram feitas para trabalhar juntas.

# Tipos de gráficos com Plotly:
    Gráfico de linha (line)

    Gráfico de barras (bar)

    Gráfico de pizza (pie)

    Dispersão (scatter)

    Mapas geográficos, 3D, entre outros.


# Como Dash e Plotly se conectam?
Plotly cria os gráficos.

Dash cria a interface da web e exibe os gráficos dentro dela.

Quando você cria um gráfico com plotly.express, pode colocá-lo dentro de um dcc.Graph() no Dash.
