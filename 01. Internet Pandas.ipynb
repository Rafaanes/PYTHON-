### Caso 2: CSV - Web

OBJETIVO:
Este programa baixa da internet uma tabela
com os preços históricos do café (do site Cooxupé),
limpa o texto (tirando aspas) e salva o resultado
em um arquivo Excel/CSV limpo, pronto para uso.

site: https://portalweb.cooxupe.com.br:9080/portal/precohistoricocafe_2.jsp

```python
# ------------------------------------------
# Importa as bibliotecas necessárias
# ------------------------------------------

# pandas: usada para ler, tratar e salvar tabelas de dados
import pandas as pd

# requests: usada para acessar sites e baixar arquivos
import requests 

# io: permite transformar textos em "arquivos virtuais" 
# (sem precisar salvar no computador)
import io 

# ============================
# 1) Baixar o conteúdo do site
# ============================
# O link abaixo é da Cooxupé, que publica preços históricos do café.
# O comando requests.get(url) baixa o conteúdo do site.
# .content retorna o conteúdo bruto (em bytes).
url = 'https://portalweb.cooxupe.com.br:9080/portal/precohistoricocafe_2.jsp;jsessionid=2CFF42FD6C9B3818526624E9811991C2?d-3496238-e=2&6578706f7274=1'

conteudo = requests.get(url).content

# ============================
# 2) Converter para texto legível
# ============================
# Aqui transformamos o conteúdo baixado (bytes) em texto normal (string).
# O parâmetro 'latin1' garante que os acentos do português sejam lidos corretamente.
texto = conteudo.decode('latin1')

# ============================
# 3) Remover todas as aspas antes de ler
# ============================
# Alguns arquivos vêm com aspas (") em volta de cada valor.
# O comando abaixo tira todas as aspas do texto.
texto_limpo = texto.replace('"','')

# ============================
# 4) Transformar em arquivo virtual
# ============================
# O pandas precisa ler um "arquivo".
# Então, usamos io.StringIO() para criar um arquivo virtual a partir do texto.
arquivo = io.StringIO(texto_limpo)

# ============================
# 5) Ler a tabela com pandas
# ============================
# Agora o pandas lê o conteúdo como uma tabela.
# sep='\t' indica que as colunas estão separadas por TAB (tabulação).
# engine='python' é uma forma de leitura mais flexível.
cafe_df = pd.read_csv(arquivo, sep = '\t', engine = 'python')

# Mostra as 5 primeiras linhas da tabela na tela
display(cafe_df.head(5))

# ============================
# 6) Salvar em arquivo CSV limpo
# ============================
# Aqui salvamos a tabela em um arquivo CSV (tipo Excel) chamado "preco_cafe.csv".
# sep=";" define o ponto e vírgula como separador de colunas.
# index=False evita criar uma coluna extra de índice.
# encoding="latin1" mantém os acentos corretamente.
cafe_df.to_csv('preco_cafe.csv', sep = ';', index = False, encoding = 'latin1')

# Mensagem final para o usuário
display('Fim do processamento!')


```

Caso 3 - Enunciado do Exercício

<u>Objetivo:</u><br>
Baixar e ler a tabela de preços históricos do café da Cooxupé diretamente da web, limpar os dados e disponibilizá-los em formato de tabela para análise.

<u>Descrição da Tarefa:</u>

1) Acesse a página da Cooxupé que contém os preços históricos do café:
https://portalweb.cooxupe.com.br:9080/portal/precohistoricocafe_2.jsp

2) Baixe o conteúdo da página utilizando Python.

3) Converta o conteúdo para um formato legível pelo pandas.

4) Remova todas as aspas (") que possam estar no texto, deixando os dados limpos.

5) Crie um DataFrame pandas com os dados da tabela.

6) Mostre as primeiras linhas da tabela para verificação.

<u>Requisitos:</u><br>
A) Utilizar bibliotecas: pandas, requests, io.

B) Garantir que os acentos e caracteres especiais sejam lidos corretamente.

C) A tabela final deve estar limpa, sem aspas extras, pronta para análise.

```python
import pandas as pd   
# biblioteca para baixar conteúdo da web    
import requests           

# ============================
# PASSO 1: Definir a URL da página
# ============================
# Esta é a página HTML que contém a tabela com os preços históricos do café
url = 'https://portalweb.cooxupe.com.br:9080/portal/precohistoricocafe_2.jsp'

# ============================
# PASSO 2: Baixar o conteúdo da página
# ============================
# requests.get(url) envia uma "requisição" para o site
# .content retorna todo o conteúdo da página HTML em bytes
pagina = requests.get(url).content

# ============================
# PASSO 3: Extrair todas as tabelas do HTML
# ============================
# pandas.read_html() consegue encontrar todas as tabelas dentro do HTML
# Retorna uma lista de DataFrames (uma tabela = um DataFrame)
tabelas = pd.read_html(pagina) 

# ============================
# PASSO 4: Selecionar a tabela principal
# ============================
# Explicando cafe_df = tabelas[0]
# tabelas → é a lista com todas as tabelas encontradas na página.
# [0] → significa “pegar o primeiro elemento da lista”, que é geralmente a tabela principal que queremos.
# cafe_df → é o DataFrame resultante, ou seja, a tabela organizada em linhas e colunas que podemos manipular com o pandas.
# Resumo:
# tabelas = [df_cafe, df_outra, df_extra] é uma lista
# tabelas[0] → pega df_cafe (primeira tabela)
# tabelas[1] → pega df_outra (segunda tabela)
# tabelas[2] → pega df_extra (terceira tabela)
cafe_df = tabelas [0]

######### PROFESSOR:
# Explicar as diferenças de ler dados da WEB(HTML) sem formatação
# e com formatação a seguir
#display (cafe_df)

# ============================
# PASSO 5: FORMATAR NÚMEROS NO PADRÃO BRASILEIRO
# ============================================
# Converte todas as colunas (exceto 'Ano') em string formatada
# Ponto como separador de milhar e vírgula como separador decimal
# percorre todas as colunas da tabela
for col in cafe_df.columns:
        if col != 'Ano':
              
        
        # cafe_df[col].apply(lambda valor: ...) → aplica uma função a cada valor da coluna
        # lambda valor: ... → define uma função simples onde 'valor' é cada número da coluna
        # f"{valor:,.2f}" → formata o número:
        #   - , → separador de milhar (temporário)
        #   - .2f → 2 casas decimais
        #   Exemplo: 2256.79 → "2,256.79" (temporário)
        # replace(",", "X").replace(".", ",").replace("X", ".") → troca os caracteres para o formato brasileiro:
        #   1) substitui todas as vírgulas temporárias por "X" para não confundir
        #   2) substitui os pontos decimais por vírgula
        #   3) substitui os "X" temporários por ponto para os milhares
        # Resultado final: 2256.79 → "2.256,79"
        
                cafe_df[col] = cafe_df[col].apply(lambda  valor: f'{valor:,.2f}').replace(',','x').replace('.',',').replace('x','.')  


# ============================
# PASSO 6: Ocultar a primeira coluna numerada 
# ============================================
# No pandas, toda tabela (DataFrame) tem um índice padrão
# que vai de 0, 1, 2, … à esquerda.  
# Esse índice não faz parte dos dados reais, mas serve 
# para identificar cada linha.
# A linha abaixo mostra toda a tabela 'cafe_df' no display do Jupyter/Notebook
# sem exibir essa coluna de índice.  
# Isso deixa a visualização mais limpa, mostrando apenas os dados reais
display(cafe_df.style.hide(axis = 'index'))

# ============================
# PASSO 7 (opcional): Salvar a tabela limpa em CSV
# ============================
# Isso permite usar os dados em Excel ou outras análises futuras
cafe_df.to_csv('preco_cafe_HTML.csv', sep = ';', index = False, encoding = 'latin1')
```
